```python
# ============================================
# CHARGEMENT DES DONNÉES
# ============================================

from pathlib import Path
import pandas as pd
import json
import pickle

print("\n" + "=" * 70)
print("📂 CHARGEMENT DES DONNÉES TFT")
print("=" * 70)

project_root = Path.cwd()
if "notebooks" in str(project_root):
    project_root = project_root.parent

tft_dir = project_root / "data" / "tft"
train_path = tft_dir / "train_data.csv"
val_path = tft_dir / "val_data.csv"
test_path = tft_dir / "test_data.csv"
config_path = tft_dir / "tft_config.json"

print(f"\n📍 Project root: {project_root}")
print(f"📍 TFT directory: {tft_dir.absolute()}")

print("\n🔍 Vérification des fichiers...")
required_files = {
    'Train data': train_path,
    'Validation data': val_path,
    'Test data': test_path,
    'Config': config_path,
}

for name, path in required_files.items():
    if path.exists():
        print(f"✅ {name}")
    else:
        print(f"❌ {name} manquant")
        raise FileNotFoundError(f"{name} non trouvé : {path}")

print("\n⏳ Chargement...")
train_df = pd.read_csv(train_path)
val_df = pd.read_csv(val_path)
test_df = pd.read_csv(test_path)

with open(config_path, 'r') as f:
    tft_config = json.load(f)

# ⚠️ Skip scalers.pkl (créé avec NumPy 2.x, incompatible avec NumPy 1.23.5)
# Les scalers ne sont pas utilisés dans l'entraînement TFT
print("\n⚠️  Scalers.pkl ignoré (incompatibilité NumPy)")

print("✅ Données chargées")
print("=" * 70)
print(f"\n📊 DATASETS:")
print(f"   Train:       {len(train_df):,} lignes")
print(f"   Validation:  {len(val_df):,} lignes")
print(f"   Test:        {len(test_df):,} lignes")
print(f"\n⚙️  CONFIG:")
print(f"   Target:              {tft_config['target']}")
print(f"   Time-varying known:  {len(tft_config['time_varying_known'])}")
print(f"   Time-varying unknown:{len(tft_config['time_varying_unknown'])}")
print("=" * 70)
```

```python
# ============================================
# PRÉPARATION DES DONNÉES ET CRÉATION DU MODÈLE
# ============================================

from pytorch_forecasting import TimeSeriesDataSet, TemporalFusionTransformer, QuantileLoss
from pytorch_forecasting.data import GroupNormalizer, NaNLabelEncoder
import numpy as np

print("\n" + "=" * 70)
print("🧹 PRÉPARATION DES DONNÉES")
print("=" * 70)

# Configuration
config = {
    'hidden_size': 160,
    'attention_head_size': 4,
    'dropout': 0.1,
    'hidden_continuous_size': 8,
    'lstm_layers': 2,
    'learning_rate': 0.001,
    'batch_size': 128,
    'max_epochs': 100,
    'gradient_clip_val': 0.1,
    'max_encoder_length': 60,
    'max_prediction_length': 1,
    'quantiles': [0.1, 0.5, 0.9]
}

# Fonction de nettoyage
def prepare_data_for_tft(df, name):
    print(f"\n⏳ Préparation de {name}...")
    df_clean = df.copy()
    initial_len = len(df_clean)
    
    # Supprimer NaN/inf dans les colonnes critiques
    critical_cols = ['time_idx', 'group_id', 'Close', 'Ticker']
    df_clean = df_clean.dropna(subset=critical_cols)
    df_clean = df_clean[~np.isinf(df_clean['Close'])]
    
    # Corriger valeurs négatives dans Close
    negative_count = (df_clean['Close'] <= 0).sum()
    if negative_count > 0:
        print(f"   ⚠️  {negative_count} valeurs Close <= 0")
        for group_id in df_clean['group_id'].unique():
            mask = df_clean['group_id'] == group_id
            group_data = df_clean.loc[mask, 'Close']
            min_positive = group_data[group_data > 0].min()
            if pd.isna(min_positive):
                min_positive = 0.01
            df_clean.loc[mask & (df_clean['Close'] <= 0), 'Close'] = min_positive
    
    # Remplir NaN dans features numériques
    numeric_cols = df_clean.select_dtypes(include=[np.number]).columns.tolist()
    numeric_cols = [col for col in numeric_cols if col not in ['time_idx', 'group_id', 'Close']]
    
    for col in numeric_cols:
        if df_clean[col].isnull().any():
            df_clean[col] = df_clean.groupby('group_id')[col].fillna(method='ffill').fillna(method='bfill')
            if df_clean[col].isnull().any():
                df_clean[col] = df_clean[col].fillna(0)
        
        if np.isinf(df_clean[col]).any():
            df_clean[col] = df_clean[col].replace([np.inf, -np.inf], np.nan)
            df_clean[col] = df_clean[col].fillna(0)
    
    print(f"   ✅ {initial_len:,} → {len(df_clean):,} lignes")
    print(f"      Close: {df_clean['Close'].min():.4f} → {df_clean['Close'].max():.4f}")
    
    return df_clean

# Nettoyer
train_df_clean = prepare_data_for_tft(train_df, "TRAIN")
val_df_clean = prepare_data_for_tft(val_df, "VALIDATION")
test_df_clean = prepare_data_for_tft(test_df, "TEST")

print("\n" + "=" * 70)
print("⚙️  CRÉATION DU MODÈLE")
print("=" * 70)

# Créer TimeSeriesDataSet
print("\n⏳ TimeSeriesDataSets...")

training = TimeSeriesDataSet(
    train_df_clean,
    time_idx='time_idx',
    target='Close',
    group_ids=['group_id'],
    max_encoder_length=config['max_encoder_length'],
    max_prediction_length=config['max_prediction_length'],
    static_categoricals=['Ticker'],
    time_varying_known_reals=tft_config['time_varying_known'],
    time_varying_unknown_reals=tft_config['time_varying_unknown'],
    target_normalizer=GroupNormalizer(
        groups=['group_id'],
        transformation='log',
        center=False
    ),
    add_relative_time_idx=True,
    add_target_scales=True,
    add_encoder_length=True,
    allow_missing_timesteps=True,
    categorical_encoders={'Ticker': NaNLabelEncoder(add_nan=True)}
)

validation = TimeSeriesDataSet.from_dataset(training, val_df_clean, predict=False, stop_randomization=True)
test_dataset = TimeSeriesDataSet.from_dataset(training, test_df_clean, predict=False, stop_randomization=True)

print(f"✅ TimeSeriesDataSets: {len(training):,} / {len(validation):,} / {len(test_dataset):,}")

# DataLoaders
print("\n⏳ DataLoaders...")

train_dataloader = training.to_dataloader(train=True, batch_size=config['batch_size'], num_workers=0)
val_dataloader = validation.to_dataloader(train=False, batch_size=config['batch_size']*2, num_workers=0)
test_dataloader = test_dataset.to_dataloader(train=False, batch_size=config['batch_size']*2, num_workers=0)

print(f"✅ DataLoaders: {len(train_dataloader)} / {len(val_dataloader)} / {len(test_dataloader)} batches")

# Modèle TFT
print("\n⏳ Modèle TFT...")

model = TemporalFusionTransformer.from_dataset(
    training,
    learning_rate=config['learning_rate'],
    hidden_size=config['hidden_size'],
    attention_head_size=config['attention_head_size'],
    dropout=config['dropout'],
    hidden_continuous_size=config['hidden_continuous_size'],
    lstm_layers=config['lstm_layers'],
    output_size=len(config['quantiles']),
    loss=QuantileLoss(quantiles=config['quantiles']),
    reduce_on_plateau_patience=4,
)

print(f"✅ Modèle TFT: {sum(p.numel() for p in model.parameters()):,} paramètres")
print(f"   LightningModule: {isinstance(model, pl.LightningModule)}")
print("=" * 70)
```

```python
# ============================================
# ENTRAÎNEMENT
# ============================================

import warnings
warnings.filterwarnings('ignore')
from pytorch_lightning.callbacks import EarlyStopping, ModelCheckpoint, LearningRateMonitor
from pytorch_lightning.loggers import TensorBoardLogger
import time

print("\n" + "=" * 70)
print("🏋️  ENTRAÎNEMENT DU MODÈLE TFT")
print("=" * 70)

models_dir = Path('models')
models_dir.mkdir(exist_ok=True)
checkpoint_dir = models_dir / 'tft_checkpoints'
checkpoint_dir.mkdir(exist_ok=True)

callbacks = [
    EarlyStopping(monitor='val_loss', patience=10, mode='min', verbose=True),
    ModelCheckpoint(dirpath=str(checkpoint_dir), filename='tft-{epoch:02d}-{val_loss:.4f}', 
                    monitor='val_loss', mode='min', save_top_k=1, verbose=True),
    LearningRateMonitor(logging_interval='epoch')
]

logger = TensorBoardLogger(save_dir=str(models_dir), name='tft_logs')

trainer = pl.Trainer(
    max_epochs=config['max_epochs'],
    gpus=0,
    gradient_clip_val=config['gradient_clip_val'],
    callbacks=callbacks,
    logger=logger,
    enable_progress_bar=True,
    enable_model_summary=True,
    log_every_n_steps=10,
)

print("\n🚀 DÉBUT DE L'ENTRAÎNEMENT")
print("⏱️  Estimation: 30-60 minutes sur CPU\n")

start_time = time.time()

try:
    trainer.fit(model, train_dataloader, val_dataloader)
    
    training_time = time.time() - start_time
    minutes = int(training_time // 60)
    seconds = int(training_time % 60)
    
    print("\n" + "=" * 70)
    print("✅ ENTRAÎNEMENT TERMINÉ")
    print("=" * 70)
    print(f"   Durée:     {minutes}m {seconds}s")
    print(f"   Val Loss:  {trainer.callback_metrics.get('val_loss', 'N/A'):.4f}")
    print(f"   Epochs:    {trainer.current_epoch + 1}")
    
    best_model_path = trainer.checkpoint_callback.best_model_path
    print(f"\n✅ Meilleur modèle: {best_model_path}")
    best_model = TemporalFusionTransformer.load_from_checkpoint(best_model_path)
    
    training_info = {
        'training_time_seconds': training_time,
        'final_epoch': trainer.current_epoch + 1,
        'best_val_loss': float(trainer.callback_metrics.get('val_loss', 0)),
        'best_model_path': str(best_model_path),
        'config': config
    }
    
    with open(models_dir / 'training_info.json', 'w') as f:
        json.dump(training_info, f, indent=2)
    
    print("💾 Infos sauvegardées")
    print("=" * 70)
    
except Exception as e:
    print(f"\n❌ ERREUR: {type(e).__name__}: {e}")
    import traceback
    traceback.print_exc()
    raise
```

```python
# ============================================
# ÉVALUATION DU MODÈLE
# ============================================

import torch
import numpy as np

print("\n" + "=" * 70)
print("📊 ÉVALUATION DU MODÈLE")
print("=" * 70)

# Prédictions sur le test set
print("\n⏳ Prédictions sur le test set...")

# Version pytorch-forecasting 0.10.3 : predict retourne une liste
# On doit utiliser predict_with_target pour avoir les valeurs réelles
actuals_list = []
preds_list = []

best_model.eval()

with torch.no_grad():
    for batch in test_dataloader:
        # Déplacer le batch sur CPU
        x = {k: v.cpu() if isinstance(v, torch.Tensor) else v for k, v in batch[0].items()}
        y = batch[1][0].cpu()  # Target (actuals)
        
        # Prédiction
        # Le modèle retourne un tuple (output, attention_weights)
        # output shape: (batch_size, prediction_length, num_quantiles)
        predictions = best_model(x)
        
        if isinstance(predictions, tuple):
            output = predictions[0]  # Prendre le premier élément du tuple
        else:
            output = predictions
        
        # Extraire la médiane (quantile 0.5, index 1 dans [0.1, 0.5, 0.9])
        pred_median = output[:, 0, 1].numpy()  # (batch_size,)
        
        # Valeurs réelles
        actual = y[:, 0].numpy()  # (batch_size,)
        
        actuals_list.append(actual)
        preds_list.append(pred_median)

# Concaténer
actuals = np.concatenate(actuals_list)
preds = np.concatenate(preds_list)

print(f"✅ Prédictions générées : {len(actuals)} échantillons")

# Calculer les métriques
mae = np.mean(np.abs(actuals - preds))
rmse = np.sqrt(np.mean((actuals - preds) ** 2))
mape = np.mean(np.abs((actuals - preds) / (np.abs(actuals) + 1e-8))) * 100

print(f"\n📈 MÉTRIQUES TEST SET:")
print(f"   MAE:   {mae:.4f}")
print(f"   RMSE:  {rmse:.4f}")
print(f"   MAPE:  {mape:.2f}%")

# Statistiques supplémentaires
r2 = 1 - (np.sum((actuals - preds)**2) / np.sum((actuals - np.mean(actuals))**2))
print(f"\n📊 STATISTIQUES SUPPLÉMENTAIRES:")
print(f"   R² Score:         {r2:.4f}")
print(f"   Mean Actual:      {np.mean(actuals):.4f}")
print(f"   Mean Predicted:   {np.mean(preds):.4f}")
print(f"   Std Actual:       {np.std(actuals):.4f}")
print(f"   Std Predicted:    {np.std(preds):.4f}")

# Sauvegarder les métriques
metrics = {
    'mae': float(mae),
    'rmse': float(rmse),
    'mape': float(mape),
    'r2': float(r2),
    'mean_actual': float(np.mean(actuals)),
    'mean_predicted': float(np.mean(preds)),
    'std_actual': float(np.std(actuals)),
    'std_predicted': float(np.std(preds)),
    'num_samples': int(len(actuals))
}

metrics_path = models_dir / 'test_metrics.json'
with open(metrics_path, 'w') as f:
    json.dump(metrics, f, indent=2)

print(f"\n💾 Métriques sauvegardées : {metrics_path}")
print("=" * 70)
```
